##Incremental Data loading using Autoloader

```python
%sql
CREATE SCHEMA IF NOT EXISTS  netflix_catalog.net_schema;
```

```python
checkpoint_location = "abfss://silver@netflixprjdl.dfs.core.windows.net/checkpoint"
```

```python

df= spark.readStream\
  .format("cloudFiles")\
  .option("cloudFiles.format", "csv")\
  .option("cloudFiles.schemaLocation", checkpoint_location)\
  .load("abfss://raw@netflixprjdl.dfs.core.windows.net")

  
```

```python
# display(df, checkpointLocation = checkpoint_location)
```

```python
df.writeStream\
  .option("checkpointLocation", checkpoint_location)\
  .trigger(processingTime='10 seconds')\
  .start("abfss://bronze@netflixprjdl.dfs.core.windows.net/netflix_titles")
```
