###Parameters


```python
dbutils.widgets.text("sourcefolder","netflix_directors")
dbutils.widgets.text("targetfolder","netflix_directors")
```

###Variables


```python
var_sourcefolder=dbutils.widgets.get("sourcefolder")
var_targetfolder=dbutils.widgets.get("targetfolder")
```

```python
df= spark.read.format("csv")\
        .option("header",True)\
        .option("inferSchema", True)\
        .load(f"abfss://bronze@netflixprjdl.dfs.core.windows.net/{var_sourcefolder}")
```

```python
df.write.mode('append')\
    .format("delta")\
    .option("path",f"abfss://silver@netflixprjdl.dfs.core.windows.net/{var_targetfolder}")\
    .save()
```
