```python
df = (
    spark.read.format("delta")
    .option("header", True)
    .option("inferSchema", True)
    .load(
        f"abfss://bronze@netflixprjdl.dfs.core.windows.net/netflix_titles")
    )
```

```python

df.write.format('delta')\
    .mode('append')\
    .option("path","abfss://silver@netflixprjdl.dfs.core.windows.net/netflix_titles")\
    .save()
```
